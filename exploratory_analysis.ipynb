```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')

print(f"Train shape: {train.shape}")
print(f"Test shape: {test.shape}")
```

```python
# Dataset Overview
print("Train columns:")
print(train.columns.tolist())
print("\nData types:")
print(train.dtypes)
print("\nMissing values:")
print(train.isnull().sum())
```

```python
# Target Variable Distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(train['demand'], bins=50, edgecolor='black', alpha=0.7, color='steelblue')
axes[0].set_title('Demand Distribution', fontsize=12)
axes[0].set_xlabel('Demand')
axes[0].set_ylabel('Frequency')

axes[1].boxplot(train['demand'])
axes[1].set_title('Demand Boxplot', fontsize=12)
axes[1].set_ylabel('Demand')

plt.tight_layout()
plt.show()

print(f"Demand statistics:")
print(f"  Mean: {train['demand'].mean():.4f}")
print(f"  Std: {train['demand'].std():.4f}")
print(f"  Min: {train['demand'].min():.4f}")
print(f"  Max: {train['demand'].max():.4f}")
```

```python
# Numerical Features Analysis
numerical_cols = ['day', 'NumberofLanes', 'Temperature', 'demand']

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    axes[i].hist(train[col], bins=30, edgecolor='black', alpha=0.7, color='coral')
    axes[i].set_title(f'{col} Distribution', fontsize=12)
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Categorical Features Analysis
categorical_cols = ['RoadType', 'LargeVehicles', 'Landmarks', 'Weather']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    train[col].value_counts().plot(kind='bar', ax=axes[i], color='skyblue', edgecolor='black')
    axes[i].set_title(f'{col} Distribution', fontsize=12)
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Correlation Matrix
numerical_train = train.select_dtypes(include=[np.number])
corr_matrix = numerical_train.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f', square=True)
plt.title('Correlation Matrix', fontsize=14)
plt.show()

# Correlation with demand
corr_with_demand = corr_matrix['demand'].sort_values(ascending=False)
print("\nCorrelation with demand:")
print(corr_with_demand)
```

```python
# Temporal Analysis - Demand by Hour
train['hour'] = train['timestamp'].str.split(':').str[0].astype(int)
hourly_demand = train.groupby('hour')['demand'].agg(['mean', 'std']).reset_index()

plt.figure(figsize=(14, 5))
plt.plot(hourly_demand['hour'], hourly_demand['mean'], marker='o', linewidth=2, markersize=6, color='darkgreen')
plt.fill_between(hourly_demand['hour'], 
                 hourly_demand['mean'] - hourly_demand['std'],
                 hourly_demand['mean'] + hourly_demand['std'],
                 alpha=0.3, color='lightgreen')
plt.title('Demand by Hour of Day', fontsize=14)
plt.xlabel('Hour', fontsize=12)
plt.ylabel('Demand', fontsize=12)
plt.xticks(range(0, 24))
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Demand by Day
daily_demand = train.groupby('day')['demand'].mean().reset_index()

plt.figure(figsize=(14, 4))
plt.bar(daily_demand['day'], daily_demand['demand'], edgecolor='black', color='teal', alpha=0.7)
plt.title('Average Demand by Day', fontsize=14)
plt.xlabel('Day', fontsize=12)
plt.ylabel('Average Demand', fontsize=12)
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
# Weather vs Demand
plt.figure(figsize=(10, 5))
weather_demand = train.groupby('Weather')['demand'].mean().sort_values()
weather_demand.plot(kind='bar', color='purple', edgecolor='black', alpha=0.7)
plt.title('Average Demand by Weather Condition', fontsize=14)
plt.xlabel('Weather', fontsize=12)
plt.ylabel('Average Demand', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
# RoadType vs Demand
plt.figure(figsize=(10, 5))
road_demand = train.groupby('RoadType')['demand'].mean().sort_values()
road_demand.plot(kind='bar', color='orange', edgecolor='black', alpha=0.7)
plt.title('Average Demand by Road Type', fontsize=14)
plt.xlabel('Road Type', fontsize=12)
plt.ylabel('Average Demand', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```
